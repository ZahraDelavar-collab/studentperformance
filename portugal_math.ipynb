مدلسازی با هوش مصنوعی

```python
import sklearn
```

```python

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
sns.pairplot(data=portugal_math_grade,x_vars=["studytime","freetime","traveltime"] , y_vars=["G3"])
```

```python
x1=portugal_math_grade[["studytime"]]
y1=portugal_math_grade['G3']

x1_train , x1_test , y1_train , y1_test = train_test_split(x1,y1 , test_size=0.2)
```

```python
model1=LinearRegression()
```

```python
model1.fit(x1_train , y1_train)
```

```python
print(f"Model1: {round(model1.intercept_,3)} + {round(model1.coef_[0])}*studytime")
```

```python
y_predict=model1.predict(x1_test)
```

```python
plt.figure(figsize=(7,7))
plt.scatter(x=x1_test, y=y1_test, color = "purple")
plt.plot(x1_test, y_predict, color = "grey")
sns.despine()
plt.title("G3 and studytime", color = "grey", loc="left")
plt.xlabel("studytime", color = "grey", loc="left")
plt.ylabel("G3", color = "grey", loc="bottom")
plt.grid()
plt.show()
```

```python
from sklearn.metrics import r2_score , mean_squared_error
```

```python
print(f"R2: {round(r2_score(y_true=y1_test, y_pred=y_predict),2)*100}")
print(f"MSE: {round(mean_squared_error(y_true=y1_test, y_pred=y_predict),2)}")
```

```python
x2=portugal_math_grade[["freetime"]]
y2=portugal_math_grade['G3']

x2_train , x2_test , y2_train , y2_test = train_test_split(x2,y2 , test_size=0.2)
```
